### content_block
一句话解释:作为一种中间媒介，统一多模态(text、image、audio、video等)上传的格式给各模型。原理就是Langchain框架帮你解析为模型供应商所需的格式。


```python
from langchain_core.messages import HumanMessage

HumanMessage(
    content_blocks=[
        {'type': 'text', 'text': '这张图里有什么？'},
        {
            'type': 'image',
            'base64': "base64_image",   # 统一字段,对应的图片base64编码
            'mime_type': 'image/png'  # 统一字段
        }#mime:Multipurpose Internet Mail Extension多媒体类型
    ]
)
```
