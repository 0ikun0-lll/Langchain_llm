## SummarizationMiddleware
- 用于对输入的文本进行摘要


```python
from langchain.agents.middleware import SummarizationMiddleware
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain.chat_models import init_chat_model
from langchain_core.messages import AIMessage, HumanMessage
from rich import print as rprint

load_dotenv(override=True)

model = init_chat_model(
    model = "deepseek:deepseek-v4-flash",
    profile={"max_input_tokens": 128_000}
)

agent = create_agent(
    model = model,
    middleware=[
        SummarizationMiddleware(
            model = model,# 摘要使用的模型
            # 触发摘要的条件(根据token数、模型上下文大小比例、消息数),可多选
            trigger=[("tokens", 100),("fraction", 0.001),("messages",6)],
            keep=("messages", 2),
            # 摘要提示词,保留{messages}占位符用于插入消息
            summary_prompt="请对以下内容进行摘要：{messages}",
        )
    ]
)

message = [HumanMessage(content = "你好"),
           AIMessage(content = "你好，我是小李"),
           HumanMessage(content = "你是小李我是谁"),
           AIMessage(content = "你是蔡徐坤"),
           HumanMessage("你个小黑子"),
           AIMessage(content = "略略略"),
           HumanMessage("你什么意思?")]

response = agent.invoke({
    "messages": message
})

rprint(response)
```
