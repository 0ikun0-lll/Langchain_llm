## 文本切分器
### text_splitter基类
- split_text(text: str) -> list[str]:文本切分成字符串列表
- create_documents(texts: list[str], metadatas: list[dict[Any, Any]] | None = None) -> list[Document]:字符串列表切分后转为Document列表
- split_documents(documents: Iterable[Document]) -> list[Document]:提取Document内容切分后转为Document列表
### CharacterTextSplitter

```python
from langchain_text_splitters import CharacterTextSplitter

text = """
LangChain 是一个用于开发由语言模型驱动的应用程序的框架的。它提供了一套工具和抽象，使开发者能够更容易地构建复杂的应用程序。
"""

splitter = CharacterTextSplitter(
    separator="",#最优先，空字符则按字符切分
    chunk_size=20,#分块大小
    chunk_overlap=5,#分块重叠大小
    add_start_index=True,#是否添加开始索引，存在Document对象metadate里
)

texts = splitter.split_text(text)

for tex in texts:
    print(tex)

docs = splitter.create_documents([text])
for doc in docs:
    print(doc)
```

### RecursiveCharacterTextSplitter

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text="LangChain框架特性\n\n多模型集成(GPT/Claude)\n记忆管理功能\n链式调用设计。文档分析场景示例：需要处理PDF/Word等格式。"

splitter = RecursiveCharacterTextSplitter(
    separators=["\n"],#默认模式,超出分块大小时从左到右按顺序切分,超出后无更小方式则完成
    chunk_size=10,
    chunk_overlap=5,
)

texts = splitter.split_text(text)
for i, tex in enumerate(texts):
    print(f"chunk_{i}: {tex}")
```
