### TypedDict
是一种类型提示的字典类型,用于定义字典的键和值的类型。
Annotated:用于添加额外的元数据到类型,如描述、默认值等。
使用方法:name : Annotated[type,...,description]


```python
import os
import dotenv
from langchain_deepseek import ChatDeepSeek

dotenv.load_dotenv(override=True)

DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = ChatDeepSeek(
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    model="deepseek-v4-flash",
    #pydantic底层调用functioncalling,需要禁用思考模式
    #extra_body={"thinking": {"type": "disabled"}}
)
```

```python
from typing_extensions import TypedDict,Annotated


class Movie(TypedDict):
    """电影"""
    title: Annotated[str,...,"电影标题"]
    director: Annotated[list[str],...,"电影导演"]
    revenue: Annotated[float,...,"电影票房"]
    rating: Annotated[float,...,"电影评分"]

response = model.with_structured_output(Movie).invoke("请推荐一部电影")

print(response)
print(type(response))
```
