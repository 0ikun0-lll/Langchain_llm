## 参数
举例:

```python
from langchain.chat_models import init_chat_model
import os
import dotenv

dotenv.load_dotenv(override= True)
DEEPSEEK_API_KEY=os.getenv("DEEPSEEK_API_KEY")
DEEPSEEK_BASE_URL=os.getenv("DEEPSEEK_BASE_URL")

llm_deepseek = init_chat_model(
    model_provider= "deepseek",# 模型提供方,默认会根据model自动选择,调用的是对应Chat类
    api_key= DEEPSEEK_API_KEY,
    base_url= DEEPSEEK_BASE_URL,
    model= "deepseek-v4-flash",
    temperature= 0.5,# 温度参数，默认0.7,控制模型的随机性,Langchain的Temperature区间为[0,2],0表示完全确定性，2表示完全随机性
    max_tokens= 10,#最大输出token数,达到这个数后,模型会停止生成
)
```
