## 扩展
### 1.美化输出
- 可以使用rich库的print函数,会更美观
- 可以调用response.pretty_print()

### 2.profile
- ChatXXX类创建的对象可调用该方法,返回对应模型能力参数
- init_chat_model方法实例化对象也可调用该方法

### 3.查看所有参数
- model.model_fields()方法可查看模型的所有参数,返回一个字典,包含模型的所有参数

### 4.model_kwargs , extra_body
- model_kwargs参数存放那些OpenAI Compatible API支持，但LangChain没有直接列出的字段，如用于支持Function Call的tools字段
- extra_body参数存放模型厂商基于OpenAI API协议扩展的字段,如deepseek的thinking等

### 5.invoke,stream等的config.configurable
- 可以个性化配置模型调用的参数,如temperature,max_tokens等
```python
deepseek_llm.invoke(
    "你好",
    config={
        "run_name": "...", # 在LangSmith中这次运行会显示为指定名称
        "tags": ["test", "development"], # 打上标签便于分类查找
        "metadata": {"user_id": "123"}, # 记录用户ID
        "callbacks": [custom_handler], # 启用自定义回调函数
        "configurable":{
            "run_name": "...", # 在LangSmith中这次运行会显示为指定名称
            "tags": ["test", "development"], # 打上标签便于分类查找
            "metadata": {"user_id": "123"}, # 记录用户ID
            "callbacks": [custom_handler], # 启用自定义回调函数
            "configurable":{
                "model": "deepseek-reasoner", # 配置模型参数
                "temperature": 0.7, # 配置温度参数
                "max_tokens": 100 # 配置最大令牌数
                }
            }
        }
    )
```
